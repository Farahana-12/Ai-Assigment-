```python
import pandas as pd
import heapq

# Load the dataset
df = pd.read_csv("indian-cities-dataset.csv")
```

```python
# Create an empty graph
graph = {}

# Build the graph
for _, row in df.iterrows():
    origin = row["Origin"].strip()
    destination = row["Destination"].strip()
    distance = float(row["Distance"])

    if origin not in graph:
        graph[origin] = []

    if destination not in graph:
        graph[destination] = []

    graph[origin].append((destination, distance))
    graph[destination].append((origin, distance))

print("Graph created successfully!")
print("Total cities:", len(graph))
```

```python
# Dijkstra's algorithm
def dijkstra(graph, start, goal):
    pq = [(0, start)]
    distances = {start: 0}
    previous = {}

    while pq:
        current_distance, current_city = heapq.heappop(pq)

        if current_distance > distances[current_city]:
            continue

        if current_city == goal:
            break

        for neighbor, road_distance in graph.get(current_city, []):
            new_distance = current_distance + road_distance

            if neighbor not in distances or new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current_city
                heapq.heappush(pq, (new_distance, neighbor))

    if goal not in distances:
        return None, float("inf")

    path = []
    city = goal

    while city != start:
        path.append(city)
        city = previous[city]

    path.append(start)
    path.reverse()

    return path, distances[goal]
```

```python
start = input("Enter starting city: ").strip()
goal = input("Enter destination city: ").strip()

if start not in graph or goal not in graph:
    print("Invalid city name. Please check the city names.")

else:
    path, total_distance = dijkstra(graph, start, goal)

    if path:
        print("Shortest route:", " -> ".join(path))
        print("Total distance:", total_distance, "km")
    else:
        print("No route found.")
```
