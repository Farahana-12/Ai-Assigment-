```python
import random
import heapq

# Grid size
size = 70

# Obstacle density
density = 0.20

# Create grid
grid = []

for i in range(size):
    row = []
    for j in range(size):
        if random.random() < density:
            row.append("X")       # obstacle
        else:
            row.append(".")       # free space
    grid.append(row)

sx = int(input("Enter start X: "))
sy = int(input("Enter start Y: "))

gx = int(input("Enter goal X: "))
gy = int(input("Enter goal Y: "))

grid[sx][sy] = "S"
grid[gx][gy] = "G"


# Heuristic
def heuristic(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])


# A* algorithm
def astar(start, goal):

    queue = [(0, start)]
    cost = {start: 0}
    parent = {}

    while queue:

        _, current = heapq.heappop(queue)

        if current == goal:

            path = []

            while current != start:
                path.append(current)
                current = parent[current]

            path.append(start)
            path.reverse()

            return path

        x, y = current

        moves = [
            (x+1, y),
            (x-1, y),
            (x, y+1),
            (x, y-1)
        ]

        for nx, ny in moves:

            if nx < 0 or nx >= size or ny < 0 or ny >= size:
                continue

            if grid[nx][ny] == "X":
                continue

            new_cost = cost[current] + 1
            next_node = (nx, ny)

            if next_node not in cost:

                cost[next_node] = new_cost

                priority = new_cost + heuristic(next_node, goal)

                heapq.heappush(queue, (priority, next_node))

                parent[next_node] = current

    return None


# Finding path
start = (sx, sy)
goal = (gx, gy)

path = astar(start, goal)

if path:

    for x, y in path:
        if grid[x][y] != "S" and grid[x][y] != "G":
            grid[x][y] = "*"

    print("\nUGV Path:")

    for row in grid:
        print(" ".join(row))

    print("\nPath length:", len(path) - 1)
    print("Goal reached: Yes")

else:
    print("No path found.")
```
