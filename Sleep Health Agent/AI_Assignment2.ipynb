```python
import pandas as pd

df = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
```

```python
factors = [
    'Sleep Duration',
    'Quality of Sleep',
    'Stress Level',
    'Physical Activity Level',
    'Heart Rate'
]
df[factors].head()
```

```python
df['duration_score'] = (df['Sleep Duration'] / 8 * 100).clip(upper=100)
df[['Sleep Duration', 'duration_score']].head()
```

```python
df['quality_score'] = df['Quality of Sleep'] / 10 * 100
df[['Quality of Sleep', 'quality_score']].head()
```

```python
df['stress_score'] = (10 - df['Stress Level']) / 9 * 100
df[['Stress Level', 'stress_score']].head()
```

```python
df['activity_score'] = (df['Physical Activity Level'] / 60 * 100).clip(upper=100)
df[['Physical Activity Level', 'activity_score']].head()
```

```python
df['heart_rate_score'] = (
    (100 - df['Heart Rate']) / 40 * 100
).clip(lower=0, upper=100)
df[['Heart Rate', 'heart_rate_score']].head()
```

```python
df['sleep_score'] = (
    0.30 * df['duration_score'] +
    0.30 * df['quality_score'] +
    0.15 * df['stress_score'] +
    0.15 * df['activity_score'] +
    0.10 * df['heart_rate_score']
)
df[['Sleep Duration',
    'Quality of Sleep',
    'Stress Level',
    'Physical Activity Level',
    'Heart Rate',
    'sleep_score']].head()
```

```python
def classify_sleep(score):
    if score >= 80:
        return "Good"
    elif score >= 60:
        return "Average"
    else:
        return "Poor"
    
df['sleep_category'] = df['sleep_score'].apply(classify_sleep)
df[['sleep_score', 'sleep_category']].head(10)
```
